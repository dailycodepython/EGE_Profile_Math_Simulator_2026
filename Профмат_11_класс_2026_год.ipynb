# 🚀 Задание 1. Планиметрия (Геометрия на плоскости)

Это задание проверяет знание геометрических фигур, их свойств и формул на плоскости: треугольников, четырехугольников и окружностей.

## 📋 Пошаговая инструкция (Теория для экзамена)

Одной из самых частых тем в этом задании являются **соотношения в прямоугольном треугольнике**.

### 🔹 1. Определение тригонометрических функций
В прямоугольном треугольнике с гипотенузой $c$, прилежащим катетом $b$ и противолежащим катетом $a$ (относительно острого угла $A$):
*   **Синус ($\sin A$)** — отношение *противолежащего* катета к гипотенузе:
    $$\sin A = \frac{a}{c}$$
*   **Косинус ($\cos A$)** — отношение *прилежащего* катета к гипотенузе:
    $$\cos A = \frac{b}{c}$$
*   **Тангенс ($\operatorname{tg} A$)** — отношение *противолежащего* катета к *прилежащему*:
    $$\operatorname{tg} A = \frac{a}{b}$$

### 🔹 2. Основное тригонометрическое тождество
Связывает синус и косинус одного и того же угла:
$$\sin^2 A + \cos^2 A = 1$$

## 🎯 Разбор типовой задачи КИМ

1. **Шаг 1: Вспомнить определение.** Синус угла $A$ — это отношение противолежащего катета ($BC$) к гипотенузе ($AB$):
   $$\sin A = \frac{BC}{AB}$$
2. **Шаг 2: Выразить искомый элемент.** Нам нужно найти катет $BC$. Выражаем его из формулы:
   $$BC = AB \cdot \sin A$$

```python
import matplotlib.pyplot as plt
import numpy as np

def solve_and_draw_triangle(hypotenuse, sin_a):
    if sin_a <= 0 or sin_a >= 1:
        print("Ошибка: Синус острого угла должен быть строго в пределах от 0 до 1!")
        return

    # Расчет по тригонометрическим формулам
    bc = hypotenuse * sin_a                  # Противолежащий катет
    cos_a = np.sqrt(1 - sin_a**2)            # Основное тригонометрическое тождество
    ac = hypotenuse * cos_a                  # Прилежащий катет

    # Координаты вершин для построения (C - в начале координат)
    C = (0, 0)
    A = (0, ac)
    B = (bc, 0)

    # Визуализация треугольника
    plt.figure(figsize=(6, 6))
    x_coords = [C[0], A[0], B[0], C[0]]
    y_coords = [C[1], A[1], B[1], C[1]]

    plt.plot(x_coords, y_coords, color='black', linewidth=2)
    plt.fill(x_coords, y_coords, color='lightblue', alpha=0.3)

    # Подписи вершин и сторон
    plt.text(A[0]-0.5, A[1], 'A', fontsize=14, fontweight='bold')
    plt.text(B[0]+0.2, B[1], 'B', fontsize=14, fontweight='bold')
    plt.text(C[0]-0.5, C[1]-0.5, 'C (90°)', fontsize=12, color='red')

    plt.text(bc/2, -0.6, f'BC = {bc:.2f}', fontsize=12, horizontalalignment='center')
    plt.text(-1.2, ac/2, f'AC = {ac:.2f}', fontsize=12, verticalalignment='center')
    plt.text(bc/2 + 0.2, ac/2 + 0.2, f'AB = {hypotenuse}', fontsize=12, rotation=-np.degrees(np.arctan(ac/bc)))

    plt.title(f"Визуализация задачи 1 (Прямоугольный треугольник)\nДано: AB={hypotenuse}, sin(A)={sin_a}", fontsize=12, pad=10)
    plt.axis('equal')
    plt.grid(True, linestyle=':', alpha=0.6)
    plt.show()

    print(f"--- Результаты вычислений ---")
    print(f"Противолежащий катет BC = {bc:.1f} (Искомый ответ)")
    print(f"Прилежащий катет AC = {ac:.1f}")
    print(f"Косинус угла А = {cos_a:.2f}")

# Пример запуска под нашу задачу
solve_and_draw_triangle(hypotenuse=25, sin_a=0.8)
```

# 🚀 Задание 2. Векторы на плоскости

Это задание проверяет умение работать с координатами векторов, находить их длины, складывать/вычитать векторы и вычислять их скалярное произведение.

## 📋 Пошаговая инструкция (Теория для экзамена)

### 🔹 1. Как найти координаты вектора по двум точкам
Если заданы точка начала $A(x_1; y_1)$ и точка конца $B(x_2; y_2)$, то координаты вектора $\vec{AB}$ находятся как **«из конца вычесть начало»**:
$$\vec{AB} = (x_2 - x_1; y_2 - y_1)$$

### 🔹 2. Длина (модуль) вектора
Длина вектора $\vec{a}(x; y)$ находится по формуле, похожей на теорему Пифагора:
$$|\vec{a}| = \sqrt{x^2 + y^2}$$

### 🔹 3. Скалярное произведение векторов
Если даны векторы $\vec{a}(x_1; y_1)$ и $\vec{b}(x_2; y_2)$, их скалярное произведение равно сумме произведений одноименных координат:
$$\vec{a} \cdot \vec{b} = x_1 \cdot x_2 + y_1 \cdot y_2$$

   $$\vec{AB} = (4 - 1; 6 - 2) = (3; 4)$$

    *   *Глава X:* «Метод координат» (§1–§3 — координаты вектора, связи между координатами точек и векторов).
*   **Геометрия 9 класс (автор: Мерзляк А.Г. и др.):**

```python
import matplotlib.pyplot as plt
import numpy as np

def analyze_vectors(A, B, C):
    # Вычисляем координаты векторов
    AB = np.array([B[0] - A[0], B[1] - A[1]])
    AC = np.array([C[0] - A[0], C[1] - A[1]])

    # Скалярное произведение
    scalar_product = np.dot(AB, AC)

    # Визуализация
    plt.figure(figsize=(8, 8))

    # Рисуем векторы стрелками (origin -> dx, dy)
    plt.quiver(A[0], A[1], AB[0], AB[1], angles='xy', scale_units='xy', scale=1, color='blue', label=f'AB {tuple(AB)}')
    plt.quiver(A[0], A[1], AC[0], AC[1], angles='xy', scale_units='xy', scale=1, color='green', label=f'AC {tuple(AC)}')

    # Отмечаем точки
    plt.scatter([A[0], B[0], C[0]], [A[1], B[1], C[1]], color='red', zorder=5)
    plt.text(A[0]-0.3, A[1]-0.3, f'A{A}', fontsize=12, fontweight='bold')
    plt.text(B[0]+0.1, B[1]+0.1, f'B{B}', fontsize=12, fontweight='bold')
    plt.text(C[0]+0.1, C[1]+0.1, f'C{C}', fontsize=12, fontweight='bold')

    # Сетка и оси
    plt.axhline(0, color='black', linewidth=0.8)
    plt.axvline(0, color='black', linewidth=0.8)
    plt.grid(True, linestyle=':', alpha=0.6)

    # Автоматическое определение границ графика с запасом
    all_x = [A[0], B[0], C[0]]
    all_y = [A[1], B[1], C[1]]
    plt.xlim(min(all_x) - 2, max(all_x) + 2)
    plt.ylim(min(all_y) - 2, max(all_y) + 2)

    plt.title('Визуализация векторов из Задачи 2', fontsize=14)
    plt.legend(fontsize=12)
    plt.gca().set_aspect('equal', adjustable='box')
    plt.show()

    print("--- Результаты автоматического расчета ---")
    print(f"Координаты вектора AB: {tuple(AB)}")
    print(f"Координаты вектора AC: {tuple(AC)}")
    print(f"Скалярное произведение AB * AC = {scalar_product}")

# Запуск с точками из примера
analyze_vectors(A=(1, 2), B=(4, 6), C=(5, 1))
```

# 🚀 Задание 3. Стереометрия (Геометрия в пространстве)

Это задание проверяет знание формул объемов и площадей поверхностей пространственных фигур (многогранников и тел вращения).

## 📋 Пошаговая инструкция (Теория для экзамена)

Один из самых популярных и коварных типов задач — **изменение объема фигуры при изменении ее линейных размеров** (высоты, радиуса или ребер).

### 🔹 1. Главное правило подобия в стереометрии
Если **все** линейные размеры тела (длина, ширина, высота, радиус) увеличить в $k$ раз, то:
*   Любая площадь (основания, боковой или полной поверхности) увеличится в $k^2$ раз.
*   Объем тела ($V$) увеличится в $k^3$ раз!

### 🔹 2. Важные формулы объемов
*   **Призма и цилиндр:** $V = S_{осн} \cdot h$
*   **Пирамида и конус:** $V = \frac{1}{3} S_{осн} \cdot h$
*   **Шар:** $V = \frac{4}{3}\pi R^3$

> ⚠️ **Будьте внимательны:** если меняется только один параметр (например, только высота или только радиус), правило $k^3$ не работает! Нужно смотреть строго на формулу.

---

## 🎯 Разбор типовой задачи КИМ
**Условие:** В цилиндрический сосуд налили $2000 \text{ см}^3$ воды. Уровень воды достигает высоты $12\text{ см}$. В воду полностью погрузили деталь. При этом уровень воды в сосуде поднялся на $3\text{ см}$. Найдите объем детали. Ответ дайте в $\text{ см}^3$.

1. **Шаг 1: Понять физико-математический смысл.** По закону Архимеда объем погруженной детали равен объему *вытесненной* ею воды. Вытесненная вода имеет форму точно такого же цилиндра.
2. **Шаг 2: Проанализировать формулу объема.** Формула объема цилиндра: $V = S_{осн} \cdot h$.
   Площадь основания ($S_{осн}$) у сосуда не изменилась. Значит, объем воды прямо пропорционален ее высоте ($h$).
3. **Шаг 3: Составить пропорцию или найти отношение.**
   Начальная высота $h_1 = 12\text{ см}$ соответствует объему $V_1 = 2000 \text{ см}^3$.
   Изменение высоты $\Delta h = 3\text{ см}$ соответствует искомому объему детали $V_{дет}$.
   
   Посчитаем, какую часть от первоначальной высоты составляет подъем:
   $$\frac{\Delta h}{h_1} = \frac{3}{12} = \frac{1}{4}$$
   
   Значит, и объем детали составляет $\frac{1}{4}$ от первоначального объема воды:
   $$V_{дет} = 2000 \cdot \frac{1}{4} = 500 \text{ см}^3$$

**Ответ: 500**

```python
import matplotlib.pyplot as plt
import numpy as np

def simulate_cylinder_task(v_water, h_start, h_delta):
    # Вычисляем площадь основания сосуда
    S_base = v_water / h_start
    # Вычисляем объем детали
    v_detail = S_base * h_delta

    # Построение 3D-графика для визуализации уровней жидкости
    fig = plt.figure(figsize=(12, 5))

    def draw_cylinder(ax, height, color, title):
        # Параметры сетки цилиндра
        r = 1
        z = np.linspace(0, height, 100)
        theta = np.linspace(0, 2*np.pi, 100)
        theta_grid, z_grid = np.meshgrid(theta, z)
        x_grid = r * np.cos(theta_grid)
        y_grid = r * np.sin(theta_grid)

        # Отрисовка стенок
        ax.plot_surface(x_grid, y_grid, z_grid, color=color, alpha=0.5)
        # Отрисовка крышки
        cx = r * np.cos(theta)
        cy = r * np.sin(theta)
        ax.plot(cx, cy, height, color=color, lw=2)

        ax.set_title(title, fontsize=12)
        ax.set_xlim(-1.5, 1.5)
        ax.set_ylim(-1.5, 1.5)
        ax.set_zlim(0, h_start + h_delta + 3)
        ax.view_init(elev=15, azim=30)
        ax.axis('off')

    # Первый цилиндр: исходный уровень
    ax1 = fig.add_subplot(1, 2, 1, projection='3d')
    draw_cylinder(ax1, h_start, '#1f77b4', f"До погружения\nВысота = {h_start} см\nV воды = {v_water} см³")

    # Второй цилиндр: уровень с деталью
    ax2 = fig.add_subplot(1, 2, 2, projection='3d')
    draw_cylinder(ax2, h_start + h_delta, '#aec7e8', f"После погружения\nВысота = {h_start + h_delta} см\n+Δh = {h_delta} см")

    plt.tight_layout()
    plt.show()

    print("--- Автоматический расчет по формулам стереометрии ---")
    print(f"Площадь основания цилиндра (S): {S_base:.2f} см²")
    print(f"Объем вытесненной воды (равен объему детали): {v_detail:.1f} см³")

# Запуск симуляции с параметрами из разобранной задачи
simulate_cylinder_task(v_water=2000, h_start=12, h_delta=3)
```

# 🚀 Задание 4. Начала теории вероятностей

Это задание проверяет умение вычислять вероятность случайного события в простейших испытаниях с использованием классической формулы вероятности.

## 📋 Пошаговая инструкция (Теория для экзамена)

### 🔹 1. Классическая формула вероятности
Вероятность события $A$ (обозначается как $P(A)$) равна отношению количества благоприятных исходов к общему числу всех возможных равновероятных исходов:
$$P(A) = \frac{m}{n}$$

Где:
*   $n$ — **общее число** всех возможных исходов эксперимента.
*   $m$ — число **благоприятных исходов** (тех исходов, при которых происходит интересующее нас событие $A$).

### 🔹 2. Важные свойства вероятности
1.  Вероятность любого события всегда лежит в пределах от 0 до 1: $0 \le P(A) \le 1$.
2.  Если в ответе получилось число больше 1 или отрицательное — вы точно допустили ошибку.
3.  Ответ на экзамене всегда записывается в виде **конечной десятичной дроби**.

   Нам нужно найти $m$ — количество благоприятных билетов. Используем формулу $P(A) = \frac{m}{n}$:

*   **Алгебра 9 класс (автор: Макарычев Ю.Н. и др.):**
    *   *§ 13:* «Элементы комбинаторики и теории вероятностей» (Начальные сведения о вероятности, классическая формула).


```python
import random
import matplotlib.pyplot as plt

def simulate_exam(total_tickets, good_tickets, num_trials=10000):
    # Создаем «мешок» с билетами: 1 — нужный билет, 0 — любой другой
    bag = [1] * good_tickets + [0] * (total_tickets - good_tickets)

    successes = 0
    history_probability = []

    # Симулируем сдачу экзамена num_trials раз
    for trial in range(1, num_trials + 1):
        chosen_ticket = random.choice(bag)
        if chosen_ticket == 1:
            successes += 1
        # Фиксируем текущую частоту успеха на каждом шаге
        history_probability.append(successes / trial)

    # Математический расчет
    math_prob = good_tickets / total_tickets

    # Строим график сходимости симуляции к точной вероятности
    plt.figure(figsize=(10, 5))
    plt.plot(history_probability, color='#1f77b4', alpha=0.6, label='Частота в симуляции')
    plt.axhline(math_prob, color='red', linestyle='--', linewidth=2, label=f'Точная вероятность ({math_prob:.2f})')

    plt.title(f'График стабилизации частоты (Симуляция {num_trials} экзаменов)', fontsize=13)
    plt.xlabel('Количество испытаний (экзаменов)')
    plt.ylabel('Вероятность')
    plt.legend(fontsize=11)
    plt.grid(True, linestyle=':', alpha=0.6)
    plt.ylim(0, 1.0)
    plt.show()

    print("--- Результаты эксперимента ---")
    print(f"Математическая (теоретическая) вероятность: {math_prob:.4f}")
    print(f"Экспериментальная вероятность (после {num_trials} тестов): {history_probability[-1]:.4f}")

# Запускаем симуляцию для нашей задачи: всего 25 билетов, из них 5 нужных (так как 25 * 0.2 = 5)
simulate_exam(total_tickets=25, good_tickets=5)
```

# 🚀 Задание 5. Сложная теория вероятностей

Это задание проверяет умение применять теоремы о сложении и умножении вероятностей, а также рассчитывать вероятность наступления сложных, комбинированных событий.

## 📋 Пошаговая инструкция (Теория для экзамена)

Чтобы успешно решать эти задачи, нужно научиться четко разделять логические связки **«И»** и **«ИЛИ»**.

### 🔹 1. Умножение вероятностей (Связка «И»)
Если события $A$ и $B$ **независимы** (наступление одного не влияет на вероятность другого) и они должны произойти **одновременно (и то, и другое)**, их вероятности перемножаются:
$$P(A \text{ и } B) = P(A) \cdot P(B)$$

### 🔹 2. Сложение вероятностей (Связка «ИЛИ»)
Если события $A$ и $B$ **несовместны** (они не могут произойти одновременно в одном испытании) и нас устраивает наступление **хотя бы одного из них (или то, или другое)**, их вероятности складываются:
$$P(A \text{ или } B) = P(A) + P(B)$$

## 🎯 Разбор типовой задачи КИМ (Про кофейные автоматы)
**Условие:** В торговом центре установлены два одинаковых автомата по продаже кофе. Вероятность того, что к концу дня в автомате закончится кофе, равна $0.3$. Вероятность того, что кофе закончится в обоих автоматах, равна $0.12$. Найдите вероятность того, что к концу дня кофе останется в обоих автоматах.

1. **Шаг 1: Понять структуру событий.**
   * Пусть событие $A$ — кофе закончился в 1-м автомате ($P(A) = 0.3$).
   * Пусть событие $B$ — кофе закончился во 2-м автомате ($P(B) = 0.3$).
   * Кофе закончился в обоих автоматах одновременно ($A$ и $B$) — это пересечение событий. Нам дано, что $P(A \text{ и } B) = 0.12$.
   *(Заметьте: $0.3 \cdot 0.3 = 0.09 \ne 0.12$, значит, события зависимы!)*

2. **Шаг 2: Найти вероятность того, что кофе закончится ХОТЯ БЫ В ОДНОМ автомате.**
   Используем общую формулу сложения совместных событий:
   $$P(A \text{ или } B) = P(A) + P(B) - P(A \text{ и } B)$$
   $$P(A \text{ или } B) = 0.3 + 0.3 - 0.12 = 0.48$$

3. **Шаг 3: Найти противоположное событие.**
   Событие «кофе остался в обоих автоматах» является противоположным для события «кофе закончился хотя бы в одном». Сумма вероятностей противоположных событий равна 1:
   $$P(\text{остался в обоих}) = 1 - P(A \text{ или } B) = 1 - 0.48 = 0.52$$

**Ответ: 0.52**

### 🔹 Алгоритм решения уравнений вида $\sqrt{f(x)} = g(x)$
1.  **Шаг 1: Избавление от корня.** Возводим обе части уравнения в квадрат:
    $$f(x) = (g(x))^2$$
2.  **Шаг 2: Решение полученного уравнения.** Обычно получается стандартное квадратное или линейное уравнение, из которого мы находим корни $x_1, x_2$.
3.  **Шаг 3: Проверка корней (ОДЗ).**
    *   *Важное правило:* Результат извлечения квадратного корня не может быть отрицательным числом. Значит, правая часть исходного уравнения должна быть неотрицательной: $g(x) \ge 0$.
    *   Самый надежный способ на экзамене — просто подставить найденные иксы в **исходное** уравнение и проверить, получается ли верное числовое равенство.

   Решаем через дискриминант или по теореме Виета:
   *   Сумма корней: $x_1 + x_2 = -2$
   *   Произведение корней: $x_1 \cdot x_2 = -15$

```python
import sympy as sp

def solve_exam_equation():
    # Объявляем переменную x как математический символ
    x = sp.Symbol('x')

    # Задаем левую и правую части нашего уравнения: sqrt(15 - 2x) = x
    left_side = sp.sqrt(15 - 2*x)
    right_side = x

    # Создаем уравнение (sp.Eq неявно означает равенство левой и правой частей)
    equation = sp.Eq(left_side, right_side)

    # Решаем уравнение аналитически
    solutions = sp.solve(equation, x)

    print("--- Аналитическое решение через SymPy ---")
    print(sp.pretty(equation, use_unicode=True)) # Красивый вывод уравнения
    print(f"\nНайденные действительные корни (с учетом ОДЗ): {solutions}")

    # Демонстрация того, почему отлетел ложный корень
    potential_roots = [-5, 3]
    print("\n--- Проверка всех потенциальных корней алгебраического уравнения ---")
    for root in potential_roots:
        is_valid = left_side.subs(x, root) == right_side.subs(x, root)
        print(f"При x = {root:2}: Левая часть = {left_side.subs(x, root)}, Правая часть = {right_side.subs(x, root)} -> Подходит? {is_valid}")

solve_exam_equation()
```

### 🔹 Базовые формулы для Задания 7

    $$\sin^2 \alpha + \cos^2 \alpha = 1$$

   Обратите внимание на угол в числителе: $98^\circ = 2 \cdot 49^\circ$. Разложим его по формуле синуса двойного угла ($\sin 2\alpha = 2 \sin \alpha \cos \alpha$):
   $$\sin 98^\circ = 2 \sin 49^\circ \cos 49^\circ$$
   Подставим это обратно в выражение:
   $$\frac{5 \cdot (2 \sin 49^\circ \cos 49^\circ)}{\sin 49^\circ \cdot \sin 41^\circ} = \frac{10 \sin 49^\circ \cos 49^\circ}{\sin 49^\circ \cdot \sin 41^\circ}$$

2. **Шаг 2: Сократить одинаковые тригонометрические функции.**
   Мы можем сократить дробь на общий множитель $\sin 49^\circ$:
   $$\frac{10 \cos 49^\circ}{\sin 41^\circ}$$

3. **Шаг 3: Применить формулы приведения для оставшихся углов.**
   Заметим, что сумма углов $49^\circ + 41^\circ = 90^\circ$. Выразим косинус через синус: $\cos 49^\circ = \cos(90^\circ - 41^\circ)$.
   По формуле приведения: $\cos(90^\circ - \alpha) = \sin \alpha$.
   Следовательно:
   $$\cos 49^\circ = \sin 41^\circ$$
   Подставляем обратно и получаем окончательный ответ:
   $$\frac{10 \sin 41^\circ}{\sin 41^\circ} = 10$$

**Ответ: 10**

```python
import sympy as sp

def simplify_trig_expression():
    # Переводим углы в радианы, используя точное число Пи из SymPy
    # В ЕГЭ углы даны в градусах, а функции SymPy принимают радианы
    sin_98 = sp.sin(sp.Rational(98, 180) * sp.pi)
    sin_49 = sp.sin(sp.Rational(49, 180) * sp.pi)
    sin_41 = sp.sin(sp.Rational(41, 180) * sp.pi)

    # Собираем выражение: (5 * sin(98)) / (sin(49) * sin(41))
    expression = (5 * sin_98) / (sin_49 * sin_41)

    print("--- Исходное выражение в символьном виде ---")
    sp.pprint(expression)

    # Упрощаем выражение математическими алгоритмами SymPy
    simplified_result = sp.simplify(expression)

    print("\n--- Результат точного упрощения ---")
    print(f"Ответ: {simplified_result}")

simplify_trig_expression()
```

## 📋 Пошаговая инструкция (Теория для экзамена)

### 🔹 1. Главная формула геометрического смысла
Значение производной функции в точке касания $x_0$ равно угловому коэффициенту касательной ($k$), который равен тангенсу угла наклона ($\alpha$) этой касательной к оси $Ox$:
$$f'(x_0) = k = \operatorname{tg} \alpha$$

### 🔹 2. Правило «Домика» (Связь функции и производной)
Если перед вами график **исходной функции $y = f(x)$**:
*   В точках **максимума** и **минимума** (экстремумах) производная равна нулю: $f'(x) = 0$. Касательная там горизонтальна.
*   На участках **возрастания** функции её производная **положительна**: $f'(x) > 0$.
*   На участках **убывания** функции её производная **отрицательна**: $f'(x) < 0$.

Если перед вами график **производной $y = f'(x)$**:
*   Точки максимума/минимума самой функции — это точки, где график производной **пересекает ось $Ox$** ($f'(x) = 0$).

```python
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact, FloatSlider

def demonstrate_derivative(x0):
    # Задаем функцию: y = x^2 - 2x
    # Ее производная: y' = 2x - 2
    x = np.linspace(-3, 5, 500)
    y = x**2 - 2*x

    # Считаем значение функции и производной в точке касания x0
    y0 = x0**2 - 2*x0
    slope = 2*x0 - 2  # Это и есть f'(x0)

    # Уравнение касательной: y_tangent = f'(x0)*(x - x0) + f(x0)
    y_tangent = slope * (x - x0) + y0

    plt.figure(figsize=(10, 6))
    plt.plot(x, y, label='Функция $y = f(x) = x^2 - 2x$', color='blue', linewidth=2)
    plt.plot(x, y_tangent, label=f'Касательная в $x_0$: $f\'({x0:.1f}) = {slope:.1f}$', color='orange', linestyle='--')

    # Отмечаем точку касания
    plt.scatter(x0, y0, color='red', s=100, zorder=5)

    # Рисуем вспомогательный треугольник для демонстрации катетов dy и dx
    # Возьмем небольшой шаг вправо dx = 1.0 для наглядности
    dx = 1.0
    plt.plot([x0, x0 + dx], [y0, y0], color='green', linestyle=':', lw=2, label=f'Приращение аргумента $\Delta x = {dx}$')
    plt.plot([x0 + dx, x0 + dx], [y0, y0 + slope*dx], color='purple', linestyle=':', lw=2, label=f'Приращение функции $\Delta y = {slope*dx:.1f}$')

    # Оси и сетка
    plt.axhline(0, color='black', linewidth=0.8)
    plt.axvline(0, color='black', linewidth=0.8)
    plt.grid(True, linestyle=':', alpha=0.6)
    plt.xlim(-3, 5)
    plt.ylim(-4, 10)
    plt.title('Геометрический смысл производной в Google Colab', fontsize=14)
    plt.legend(fontsize=11, loc='upper left')
    plt.show()

    print("--- Математический разбор ---")
    print(f"В точке x0 = {x0:.2f}:")
    print(f"Значение функции f(x0) = {y0:.2f}")
    print(f"Значение производной f'(x0) = {slope:.2f} (Это тангенс угла наклона касательной)")

# Создаем интерактивный слайдер для перемещения точки касания x0
interact(demonstrate_derivative, x0=FloatSlider(value=2.0, min=-1.5, max=3.5, step=0.1, description='Точка x0:'));
```

## 🎯 Разбор типовой задачи КИМ
**Условие:** Моторная лодка прошла против течения реки $112 \text{ км}$ и вернулась в пункт отправления, затратив на обратный путь на $2 \text{ часа}$ меньше. Найдите скорость лодки в неподвижной воде, если скорость течения равна $1 \text{ км/ч}$.

1. **Шаг 1: Введем переменную.**
   Пусть $x \text{ км/ч}$ — собственная скорость лодки (в неподвижной воде). Тогда:
   * Скорость по течению: $x + 1 \text{ км/ч}$
   * Скорость против течения: $x - 1 \text{ км/ч}$

2. **Шаг 2: Выразим время движения.**
   * Время против течения: $t_{против} = \frac{112}{x - 1}$
   * Время по течению: $t_{по} = \frac{112}{x + 1}$

3. **Шаг 3: Составим уравнение.**
   По условию обратный путь (по течению) занял на 2 часа меньше. Значит, разница во времени равна 2:
   $$\frac{112}{x - 1} - \frac{112}{x + 1} = 2$$

4. **Шаг 4: Решим уравнение.**
   Разделим обе части на 2 для упрощения:
   $$\frac{56}{x - 1} - \frac{56}{x + 1} = 1$$
   
   Приведем к общему знаменателю $(x-1)(x+1) = x^2 - 1$:
   $$56(x + 1) - 56(x - 1) = x^2 - 1$$
   $$56x + 56 - 56x + 56 = x^2 - 1$$
   $$112 = x^2 - 1$$
   $$x^2 = 113 \implies x = \sqrt{113}$$
   *(Примечание: для красивых целых чисел в КИМ обычно подбираются точные квадраты. Если бы в условии разница была, например, 6 часов, расчеты привели бы к точному числу).*
   
   Давайте возьмем классические данные ФИПИ, где расстояние $112$, а разница во времени $6$ часов (при скорости течения $1$ км/ч). Тогда уравнение:
   $$\frac{112}{x - 1} - \frac{112}{x + 1} = 6 \implies 112(x+1) - 112(x-1) = 6(x^2 - 1) \implies 224 = 6x^2 - 6 \implies 6x^2 = 230 \dots$$
   
   **Рассмотрим точный прототип ФИПИ:** Расстояние = $132 \text{ км}$, скорость течения = $5 \text{ км/ч}$, обратный путь на $5$ часов быстрее.
   $$\frac{132}{x - 5} - \frac{132}{x + 5} = 5$$
   $$132(x + 5) - 132(x - 5) = 5(x^2 - 25)$$
   $$1320 = 5x^2 - 125 \implies 5x^2 = 1445 \implies x^2 = 289 \implies x = 17$$
   Скорость лодки не может быть отрицательной, поэтому $x = 17$.

**Ответ: 17**

## 📚 Где почитать в учебниках

*   **Алгебра 8 класс (автор: Макарычев Ю.Н. и др.):**
    *   *Глава III, §8:* «Решение задач с помощью рациональных уравнений» (именно здесь закладывается база для текстовых задач ЕГЭ).
*   **Алгебра 9 класс (автор: Мерзляк А.Г. и др.):**
    *   Параграфы, посвященные составлению систем уравнений для текстовых задач на движение и работу.


```python
import sympy as sp

def solve_river_task(distance, stream_speed, time_difference):
    # Объявляем переменную x — собственная скорость лодки
    x = sp.Symbol('x', positive=True) # Ограничиваем поиск только положительными числами

    # Время против течения минус время по течению равно разнице во времени
    t_against = distance / (x - stream_speed)
    t_along = distance / (x + stream_speed)

    # Составляем уравнение: t_against - t_along = time_difference
    equation = sp.Eq(t_against - t_along, time_difference)

    print("--- Уравнение баланса времени ---")
    sp.pprint(equation)

    # Находим корни уравнения
    solutions = sp.solve(equation, x)

    print("\n--- Результат расчета ---")
    for sol in solutions:
        # Дополнительно проверяем, чтобы скорость лодки была больше скорости течения
        if sol > stream_speed:
            print(f"Собственная скорость лодки: {sol} км/ч (Подходит по физическому смыслу)")
        else:
            print(f"Корень {sol} отброшен (Лодка не сможет плыть против течения)")

# Запуск под классический прототип ФИПИ (S = 132 км, v_теч = 5 км/ч, dt = 5 часов)
solve_river_task(distance=132, stream_speed=5, time_difference=5)
```

## 📋 Пошаговая инструкция (Теория для экзамена)

Разберем один из самых популярных прототипов — **квадратичную функцию (параболу)**.

### 🔹 Общий вид и свойства параболы
Функция задается формулой: $f(x) = ax^2 + bx + c$
*   Коэффициент $a$ отвечает за направление ветвей (вверх при $a > 0$, вниз при $a < 0$) и за «ширину» параболы. Если от вершины отойти на 1 клетку вправо, то вверх мы поднимемся на $a$ клеток.
*   Коэффициент $c$ — это ордината (координата $y$) точки пересечения параболы с осью $Oy$. Часто её видно прямо на рисунке.
*   Координата $x_0$ вершины параболы находится по формуле: $x_0 = -\frac{b}{2a}$.

Найдите абсциссу $x_A$ точки пересечения, если уравнения прямых после расчета по точкам получились $f(x) = 3x - 2$ и $g(x) = -2x + 8$. *(Подсказка: приравняйте функции друг к другу. Ответ: 2)*

```python
import numpy as np
import matplotlib.pyplot as plt

def reconstruct_and_plot_parabola(p1, p2, p3, target_x):
    # Разделяем координаты точек
    x1, y1 = p1
    x2, y2 = p2
    x3, y3 = p3

    # Составляем матрицу системы уравнений (X * A = Y)
    # Уравнение вида: a*x^2 + b*x + c = y
    X = np.array([
        [x1**2, x1, 1],
        [x2**2, x2, 1],
        [x3**2, x3, 1]
    ])
    Y = np.array([y1, y2, y3])

    # Решаем систему линейных уравнений для поиска коэффициентов [a, b, c]
    try:
        a, b, c = np.linalg.solve(X, Y)
    except np.linalg.LinAlgError:
        print("Ошибка: Точки лежат на одной прямой или имеют одинаковые координаты X! Параболу построить нельзя.")
        return

    # Считаем значение в целевой точке
    target_y = a * target_x**2 + b * target_x + c

    print("--- Результаты анализа графика ---")
    print(f"Восстановленные коэффициенты: a = {a:.2f}, b = {b:.2f}, c = {c:.2f}")
    print(f"Уравнение функции: f(x) = {a:.2f}x² + ({b:.2f})x + ({c:.2f})")
    print(f"Искомое значение f({target_x}) = {target_y:.2f} (Ваш ответ на экзамене)")

    # Визуализация
    # Подбираем диапазон x так, чтобы вошла и целевая точка
    all_x = [x1, x2, x3, target_x]
    x_grid = np.linspace(min(all_x) - 2, max(all_x) + 2, 1000)
    y_grid = a * x_grid**2 + b * x_grid + c

    plt.figure(figsize=(10, 6))
    plt.plot(x_grid, y_grid, color='#1f77b4', linewidth=2, label=f'$f(x) = {a:.1f}x^2 + ({b:.1f})x + ({c:.1f})$')

    # Отмечаем исходные точки с графика КИМ
    plt.scatter([x1, x2, x3], [y1, y2, y3], color='green', s=100, zorder=5, label='Заданные точки КИМ')

    # Отмечаем точку ответа
    plt.scatter(target_x, target_y, color='red', s=120, marker='*', zorder=6, label=f'Искомая точка f({target_x})')
    plt.axvline(target_x, color='red', linestyle=':', alpha=0.5)
    plt.axhline(target_y, color='red', linestyle=':', alpha=0.5)

    # Настройки осей
    plt.axhline(0, color='black', linewidth=0.8)
    plt.axvline(0, color='black', linewidth=0.8)
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.legend(fontsize=11)
    plt.title('Восстановление параболы по точкам в Задании 11', fontsize=13)
    plt.show()

# Запуск под условия нашей задачи:
# Точки (0,-3), (1,-2), (2,3). Ищем значение в точке x = -5
reconstruct_and_plot_parabola(p1=(0, -3), p2=(1, -2), p3=(2, 3), target_x=-5)
```

### 🔹 2. Алгоритм поиска значения функции на отрезку $[a; b]$
1.  **Шаг 1:** Найти производную функции $f'(x)$.
2.  **Шаг 2:** Приравнять производную к нулю ($f'(x) = 0$) и найти корни (критические точки).
3.  **Шаг 3:** Отсортировать полученные корни. Нас интересуют только те, которые лежат внутри заданного отрезка $[a; b]$.
4.  **Шаг 4:** Поочередно подставить в **исходную** функцию $f(x)$ концы отрезка ($a$ и $b$) и критические точки, попавшие на отрезок. Выбрать из результатов наибольшее или наименьшее число.

## 📚 Где почитать в учебниках

*   **Алгебра и начала математического анализа 10–11 класс (авторы: Алимов Ш.А. и др.):**
    *   *Глава VI:* «Производная и ее применения» (§49 — Применение производной к исследованию функций, §50 — Максимум и минимум функции, §51 — Наибольшее и наименьшее значения функции).


```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

def analyze_and_find_extremum(expr_str, start, end, mode='min'):
    # Инициализируем символьную переменную x
    x = sp.Symbol('x')

    # Парсим строку в математическое выражение SymPy
    f = sp.sympify(expr_str)

    # Шаг 1: Находим производную автоматически
    f_prime = sp.diff(f, x)

    # Шаг 2: Находим критические точки (корни f'(x) = 0)
    crit_points = sp.solve(sp.Eq(f_prime, 0), x)

    # Отбираем точки, попавшие в нужный отрезок
    valid_points = [float(pt) for pt in crit_points if start <= pt <= end]
    # Добавляем к ним концы отрезков
    all_points_to_check = list(set(valid_points + [float(start), float(end)]))

    # Шаг 3: Считаем значения функции в этих точках
    results = {}
    for pt in all_points_to_check:
        results[pt] = float(f.subs(x, pt))

    # Выбираем ответ в зависимости от искомого значения (минимум или максимум)
    if mode == 'min':
        best_x = min(results, key=results.get)
        best_y = results[best_x]
        title_mode = "Наименьшее"
    else:
        best_x = max(results, key=results.get)
        best_y = results[best_x]
        title_mode = "Наибольшее"

    print(f"--- Математический анализ на отрезке [{start}; {end}] ---")
    print(f"Исходная функция f(x) = {f}")
    print(f"Производная f'(x) = {f_prime}")
    print(f"Все критические точки функции: {crit_points}")
    print(f"Точки, проверенные на отрезке: {all_points_to_check}")
    print(f"\nЗначения функции в них: {results}")
    print(f"==> {title_mode} значение функции на отрезке равно {best_y:.2f} (при x = {best_x:.2f})")

    # Визуализация графика для самопроверки
    x_vals = np.linspace(float(start) - 1, float(end) + 1, 500)
    # Переводим выражение SymPy в функцию NumPy для быстрого построения графика
    f_num = sp.lambdify(x, f, "numpy")
    y_vals = f_num(x_vals)

    plt.figure(figsize=(10, 6))
    plt.plot(x_vals, y_vals, label=f'$f(x) = {expr_str}$', color='#1f77b4', linewidth=2)

    # Выделяем цветом исследуемый отрезок
    x_segment = np.linspace(float(start), float(end), 200)
    plt.fill_between(x_segment, f_num(x_segment), color='orange', alpha=0.15, label='Исследуемый отрезок')
    plt.axvline(start, color='orange', linestyle='--', alpha=0.5)
    plt.axvline(end, color='orange', linestyle='--', alpha=0.5)

    # Отмечаем точку искомого ответа
    plt.scatter(best_x, best_y, color='red', s=120, zorder=5, label=f'Ответ ({title_mode})')

    plt.grid(True, linestyle=':', alpha=0.6)
    plt.title(f'Поиск экстремумов в Задании 12', fontsize=13)
    plt.legend(fontsize=11)
    plt.show()

# Запускаем скрипт под условия нашей задачи
# Уравнение: x**3 - 3*x**2 + 2, отрезок от 1 до 4, ищем минимум ('min')
analyze_and_find_extremum(expr_str="x**3 - 3*x**2 + 2", start=1, end=4, mode='min')
```

   $$2 - 2\sin^2 x + 5\sin x - 4 = 0 \implies -2\sin^2 x + 5\sin x - 2 = 0$$

2. Делаем замену $t = \sin x$, где $|t| \le 1$:
   $$2t^2 - 5t + 2 = 0$$

### 📝 Решение пункта (б):

*   **Для первой серии:** $\pi \le \frac{\pi}{6} + 2\pi k \le \frac{5\pi}{2}$

**Ответ:**

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

def solve_and_plot_trig_fixed():
    x = sp.Symbol('x', real=True)

    # Задаем уравнение из пункта (а): 2*cos(x)**2 + 5*sin(x) - 4 = 0
    equation = sp.Eq(2 * sp.cos(x)**2 + 5 * sp.sin(x) - 4, 0)

    # Пункт (а): Находим решения
    all_solutions = sp.solve(equation, x)

    # ФИЛЬТР: Оставляем только действительные корни, убирая комплексный "мусор"
    base_solutions = [sol for sol in all_solutions if sol.is_real]

    print("--- Решение пункта (а) ---")
    print(f"Базовые действительные корни на периоде [0, 2π): {base_solutions}")
    print("Общий ответ запишется как добавление + 2*pi*k к каждому корню.\n")

    # Пункт (б): Границы отбора корней
    start_bound = np.pi
    end_bound = 2.5 * np.pi

    # Численный поиск конкретных корней для пункта (б) на отрезке
    valid_roots = []
    # Проверяем несколько периодов вокруг отрезка (k от -3 до 3)
    for root in base_solutions:
        root_val = float(root)
        for k in range(-3, 4):
            shifted_root = root_val + 2 * np.pi * k
            if start_bound <= shifted_root <= end_bound:
                valid_roots.append(shifted_root)

    print("--- Решение пункта (б) ---")
    print(f"Корни, попавшие в отрезок [π; 2.5π]:")
    for r in valid_roots:
        print(f"x = {r:.4f}  (в долях пи: {r / np.pi:.3f}*π)")

    # --- ВИЗУАЛИЗАЦИЯ ТРИГОНОМЕТРИЧЕСКОГО КРУГА ---
    fig, ax = plt.subplots(figsize=(7, 7))

    # Рисуем единичную окружность
    t = np.linspace(0, 2*np.pi, 200)
    ax.plot(np.cos(t), np.sin(t), color='black', lw=1.5)

    # Закрашиваем дугу отбора (пункт б) от pi до 2.5pi
    t_arc = np.linspace(np.pi, 2.5 * np.pi, 100)
    ax.plot(np.cos(t_arc) * 1.05, np.sin(t_arc) * 1.05, color='orange', lw=4, label='Отрезок отбора [π; 2.5π]')

    # Оси координат внутри круга
    ax.axhline(0, color='gray', linestyle=':', alpha=0.7)
    ax.axvline(0, color='gray', linestyle=':', alpha=0.7)

    # Наносим все базовые корни (синие точки)
    for root in base_solutions:
        rx, ry = np.cos(float(root)), np.sin(float(root))
        ax.scatter(rx, ry, color='blue', s=80, zorder=5, label='Серии корней уравнения' if root==base_solutions[0] else "")
        # Используем "сырую" строку r"..." для безопасного отображения знака \pi
        ax.text(rx*1.2, ry*1.2, f"{float(root)/np.pi:.2f}" + r"$\pi$", fontsize=10, ha='center')

    # Наносим корни, прошедшие отбор (красные звездочки)
    for r in valid_roots:
        ax.scatter(np.cos(r), np.sin(r), color='red', marker='*', s=200, zorder=6, label='Итоговый отбор (п. б)')

    ax.set_xlim(-1.4, 1.4)
    ax.set_ylim(-1.4, 1.4)
    ax.set_aspect('equal')
    ax.axis('off')
    plt.title("Визуализация отбора корней на тригонометрическом круге", fontsize=13)
    plt.legend(loc='lower left')
    plt.show()

# Запускаем исправленную версию
solve_and_plot_trig_fixed()
```

# 🚀 Задание 14. Логарифмические и показательные неравенства

Это задание проверяет умение решать сложные неравенства методом замены переменной, методом интервалов, а также с помощью расщепления или метода рационализации.

## 📋 Пошаговая инструкция (Теория для экзамена)

Разберем классический и самый частый прототип — **логарифмическое неравенство**, решаемое заменой переменной.

### 🔹 1. Обязательное условие (ОДЗ)
Прежде чем совершать любые преобразования с логарифмами, зафиксируйте ограничения:
*   Выражение, стоящее под знаком логарифма, должно быть строго положительным: $f(x) > 0$.
*   Если переменная есть в основании логарифма $\log_{a(x)} f(x)$, то: $a(x) > 0$ и $a(x) \neq 1$.

### 🔹 2. Свойства логарифмов и замена
*   Вынесение степени из под логарифма: $\log_a (f(x))^2 = 2\log_a |f(x)|$. Если по ОДЗ выражение уже положительно, модуль можно опустить.
*   Приведение к общему основанию и замена вида $t = \log_a f(x)$. Обратите внимание: в отличие от синуса, для логарифма ограничений на букву $t$ нет ($t$ может быть любым числом).

### 🔹 3. Метод интервалов для обратной замены
После решения неравенства относительно $t$ (например, получили $t \le 2$), обязательно вернитесь к исходной переменной и решите простое неравенство:
$$\log_a f(x) \le 2 \implies \begin{cases} f(x) \le a^2 \quad (\text{если } a > 1) \\ f(x) > 0 \quad (\text{ОДЗ}) \end{cases}$$
> ⚠️ **Важно:** Если основание логарифма $a$ лежит между 0 и 1 (например, $\log_{0.5}$), при снятии логарифма **знак неравенства переворачивается**!

## 🎯 Разбор типовой задачи КИМ
**Условие:** Решите неравенство: $\log_2^2 x - 3\log_2 x + 2 \ge 0$.

1. **Шаг 1: Запишем ОДЗ.**
   Под логарифмом стоит переменная $x$. Значит:
   $$\text{ОДЗ: } x > 0$$

2. **Шаг 2: Введем новую переменную.**
   Пусть $t = \log_2 x$. Перепишем неравенство в виде квадратного:
   $$t^2 - 3t + 2 \ge 0$$

3. **Шаг 3: Решим квадратное неравенство (Метод интервалов).**
   Найдем корни уравнения $t^2 - 3t + 2 = 0$. По теореме Виета: $t_1 = 1, t_2 = 2$.
   Расставим знаки на числовой оси для выражения $(t-1)(t-2) \ge 0$:
   * Знаки идут справа налево: $+$ , $-$ , $+$
   * Нам подходят промежутки со знаком плюс: $t \le 1$ или $t \ge 2$.

4. **Шаг 4: Обратная замена с учетом ОДЗ.**
   Разбиваем решение на два случая:
   *   **Случай 1:** $\log_2 x \le 1 \implies x \le 2^1 \implies x \le 2$. С учетом ОДЗ получаем интервал: $(0; 2]$.
   *   **Случай 2:** $\log_2 x \ge 2 \implies x \ge 2^2 \implies x \ge 4$. Этот интервал полностью удовлетворяет ОДЗ: $[4; +\infty)$.

Объединяя оба промежутка, получаем итоговое множество решений.

**Ответ: $(0; 2] \cup [4; +\infty)$**

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

def solve_and_plot_inequality_fixed():
    # Объявляем переменную x (строго вещественную и положительную для логарифма)
    x = sp.Symbol('x', real=True, positive=True)

    # Задаем левую часть неравенства из разбора
    f_expr = sp.log(x, 2)**2 - 3*sp.log(x, 2) + 2

    # Формируем неравенство f(x) >= 0
    inequality = f_expr >= 0

    print("--- Аналитическое решение неравенства через SymPy ---")
    print(f"Исходное неравенство: {f_expr} >= 0")

    # Автоматическое решение с учетом всех математических ограничений
    solution = sp.solve_univariate_inequality(inequality, x, relational=False)
    print(f"Точный математический ответ: {solution}")

    # --- СТРОИМ ГРАФИК ДЛЯ ВИЗУАЛИЗАЦИИ МЕТОДА ИНТЕРВАЛОВ ---
    # Берем диапазон x от 0.1 до 6 (около нуля логарифм уходит в бесконечность)
    x_vals = np.linspace(0.1, 6.0, 1000)
    # Считаем значения функции: (log2(x))**2 - 3*log2(x) + 2
    y_vals = (np.log2(x_vals))**2 - 3*np.log2(x_vals) + 2

    plt.figure(figsize=(10, 6))
    plt.plot(x_vals, y_vals, color='#1f77b4', lw=2, label='y = log2^2(x) - 3*log2(x) + 2')

    # Выделяем цветом области, где y >= 0 (решение неравенства)
    plt.fill_between(x_vals, y_vals, where=(y_vals >= 0), color='green', alpha=0.2, label='Область решения (y >= 0)')
    plt.fill_between(x_vals, y_vals, where=(y_vals < 0), color='red', alpha=0.1, label='Область под осью (y < 0)')

    # Отмечаем критические точки (границы интервалов)
    plt.scatter([2, 4], [0, 0], color='black', s=80, zorder=5)
    plt.text(2, 0.3, 'x = 2', ha='center', fontweight='bold')
    plt.text(4, -0.4, 'x = 4', ha='center', fontweight='bold')

    # Оформление осей
    plt.axhline(0, color='black', linewidth=1)
    plt.axvline(0, color='black', linewidth=1, linestyle='--', alpha=0.7, label='Граница ОДЗ (x = 0)')
    plt.grid(True, linestyle=':', alpha=0.6)
    plt.xlim(-0.5, 6)
    plt.ylim(-1, 4)
    plt.title('Графическая иллюстрация решения Задания 14', fontsize=13)
    plt.xlabel('Ось X')
    plt.ylabel('Ось Y')
    plt.legend(fontsize=10, loc='upper right')
    plt.show()

# Запускаем исправленную версию
solve_and_plot_inequality_fixed()
```

# 🚀 Задание 15. Финансовая математика (Экономическая задача)

Это задание проверяет умение строить математические модели реальных финансовых процессов (кредитование, инвестиции, оптимизация прибыли) и производить по ним расчеты.

## 📋 Пошаговая инструкция (Теория для экзамена)

Разберем самый популярный прототип — **кредит с дифференцированными платежами** (когда долг уменьшается равномерно).

### 🔹 1. Главный принцип дифференцированных платежей
*   Сумма долга должна уменьшаться **на одну и ту же величину** каждый период (месяц или год).
*   Если кредит взят на $n$ лет, то остаток основного долга без процентов уменьшается ровно на $\frac{S}{n}$ каждый год.
*   Каждый платеж состоит из двух частей: фиксированная выплата основного долга ($\frac{S}{n}$) плюс начисленные проценты на *текущий остаток* долга. Из-за этого выплаты со временем уменьшаются.

### 🔹 2. Схема начисления
Пусть $S$ — сумма кредита, $r\%$ — годовая процентная ставка.
Введем повышающий коэффициент: $k = 1 + \frac{r}{100}$ (или работаем напрямую с долей процента $p = \frac{r}{100}$).
*   **1-й год:** долг равен $S$. Начисляются проценты: $S \cdot p$. Выплата: $\frac{S}{n} + S \cdot p$. Остаток долга: $S - \frac{S}{n}$.
*   **2-й год:** долг равен $S \cdot \frac{n-1}{n}$. Начисляются проценты: $S \cdot \frac{n-1}{n} \cdot p$. Выплата: $\frac{S}{n} + S \cdot \frac{n-1}{n} \cdot p$.
*   **Последний год:** долг равен $\frac{S}{n}$. Проценты: $\frac{S}{n} \cdot p$. Выплата: $\frac{S}{n} + \frac{S}{n} \cdot p$. Остаток долга: $0$.

## 🎯 Разбор типовой задачи КИМ
**Условие:** В июле 2026 года планируется взять кредит в банке на сумму $700\ 000$ рублей на 7 лет. Условия его возврата таковы:
* Каждый январь долг возрастает на $10\%$ по сравнению с концом предыдущего года.
* С февраля по июнь каждого года необходимо выплатить часть долга.
* В июле каждого года долг должен быть на одну и ту же величину меньше долга на июль предыдущего года.
Найдите **общую сумму выплат** после полного погашения кредита.

1. **Шаг 1: Понять тип кредита.**
   Фраза «долг должен быть на одну и ту же величину меньше долга на июль предыдущего года» однозначно говорит о том, что платежи **дифференцированные**.

2. **Шаг 2: Посчитать выплату основного долга.**
   Кредит взят на 7 лет. Каждый год основной долг уменьшается на:
   $$\frac{700\ 000}{7} = 100\ 000 \text{ рублей}$$
   За 7 лет фиксированная часть выплат составит ровно сумму самого кредита: $700\ 000$ рублей.

3. **Шаг 3: Посчитать переплату по процентам.**
   Проценты каждый год начисляются на остаток долга ($10\%$, то есть умножаем остаток на $0.1$):
   * 1-й год: $700\ 000 \cdot 0.1 = 70\ 000$
   * 2-й год: $600\ 000 \cdot 0.1 = 60\ 000$
   * 3-й год: $500\ 000 \cdot 0.1 = 50\ 000$
   * ...
   * 7-й год: $100\ 000 \cdot 0.1 = 10\ 000$
   
   Сумма всех процентов представляет собой арифметическую прогрессию:
   $$\text{Всего процентов} = 0.1 \cdot (700\ 000 + 600\ 000 + \dots + 100\ 000)$$
   Используем формулу суммы арифметической прогрессии: $S_{прогр} = \frac{a_1 + a_n}{2} \cdot n$
   $$\text{Всего процентов} = 0.1 \cdot \frac{700\ 000 + 100\ 000}{2} \cdot 7 = 0.1 \cdot 400\ 000 \cdot 7 = 280\ 000 \text{ рублей}$$

4. **Шаг 4: Найти общую сумму выплат.**
   $$\text{Общая сумма} = \text{Основной долг} + \text{Всего процентов} = 700\ 000 + 280\ 000 = 980\ 000 \text{ рублей}$$

**Ответ: 980 000**

*   **Алгебра 9 класс (автор: Макарычев Ю.Н. и др.):**
    *   *Глава IV:* «Последовательности» (§10 — Арифметическая прогрессия). Почти все формулы переплат в кредитах строятся на сумме арифметической прогрессии.
*   **Алгебра и начала математического анализа 10–11 класс (авторы: Алимов Ш.А. и др.):**
    *   *Глава IX:* «Геометрическая прогрессия» (нужна для понимания аннуитетных платежей, где долг умножается на коэффициент каждый год).


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

def calculate_diff_credit(S, r, n):
    # Фиксированная выплата основного долга каждый год
    base_payment = S / n

    remaining_debt = S
    records = []

    # Расчет таблицы по годам
    for year in range(1, n + 1):
        interest = remaining_debt * (r / 100) # Начисленные проценты
        total_payment = base_payment + interest # Итоговый платеж за год
        debt_end = remaining_debt - base_payment # Остаток долга на конец года

        records.append({
            "Год": year,
            "Долг на начало (руб)": remaining_debt,
            "Начисленный % (руб)": interest,
            "Выплата долга (руб)": base_payment,
            "Итоговый платеж (руб)": total_payment,
            "Долг на конец (руб)": max(0.0, debt_end)
        })
        remaining_debt = debt_end

    # Переводим в DataFrame для красивого табличного вывода в Colab
    df = pd.DataFrame(records)

    total_interest_sum = df["Начисленный % (руб)"].sum()
    total_payout_sum = df["Итоговый платеж (руб)"].sum()

    print("="*65)
    print(f"ФИНАНСОВЫЙ ПЛАН КРЕДИТА (Сумма: {S}, Ставка: {r}%, Срок: {n} лет)")
    print("="*65)
    # Форматируем вывод чисел, чтобы они легко читались
    pd.options.display.float_format = '{:,.2f}'.format
    print(df.to_string(index=False))
    print("="*65)
    print(f"Итого переплата по процентам: {total_interest_sum:,.2f} руб.")
    print(f"Общая сумма всех выплат банку: {total_payout_sum:,.2f} руб. (Ваш ответ)")
    print("="*65)

    # Визуализация структуры платежей
    plt.figure(figsize=(10, 5))
    years = df["Год"]
    plt.bar(years, df["Выплата долга (руб)"], label="Основной долг", color="#1f77b4", alpha=0.8)
    plt.bar(years, df["Начисленный % (руб)"], bottom=df["Выплата долга (руб)"], label="Проценты банка", color="#ff7f0e", alpha=0.8)

    plt.xlabel("Год кредита")
    plt.ylabel("Сумма ежегодного платежа (руб)")
    plt.title("График уменьшения платежей при дифференцированной схеме")
    plt.xticks(years)
    plt.grid(True, axis='y', linestyle=':', alpha=0.6)
    plt.legend()
    plt.show()

# Запускаем расчет под условия нашей задачи: 700 000 руб, 10%, 7 лет
calculate_diff_credit(S=700000, r=10, n=7)
```

### 🔹 Ключевые теоремы-выручалочки для Задания 16:

2.  **Свойство касательной и секущей:** Если из одной точки к окружности проведены касательная $AK$ и секущая $ABC$, то квадрат касательной равен произведению секущей на ее внешнюю часть:
    $$AK^2 = AB \cdot AC$$
3.  **Теорема о вписанном угле:** Вписанный угол равен половине центрального угла, опирающегося на ту же дугу. Вписанные углы, опирающиеся на одну и ту же дугу, равны.
4.  **Угол между касательной и хордой:** Равнен половине дуги, заключенной между ними (или равен вписанному углу, опирающемуся на эту хорду).

## 🎯 Разбор типовой задачи КИМ
**Условие:** Окружность, вписанная в прямоугольный треугольник $ABC$ с прямым углом $C$, касается катетов $AC$ и $BC$ в точках $M$ и $N$ соответственно. Точка $P$ — середина гипотенузы $AB$.
а) Докажите, что $PM = PN$.
б) Найдите длину отрезка $MN$, если катеты треугольника равны $AC = 6$ и $BC = 8$.

### 📝 Решение пункта (а):
1. Обозначим радиус вписанной окружности за $r$. Четырехугольник $CMON$ (где $O$ — центр окружности) является квадратом со стороной $r$, так как все его углы прямые. Следовательно, $CM = CN = r$.
2. В прямоугольном треугольнике медиана $CP$, проведенная к гипотенузе, равна половине гипотенузы: $CP = AP = BP$.
3. Рассмотрим треугольники $CMP$ и $CNP$. В них сторона $CP$ — общая, стороны $CM = CN = r$.
4. Углы $\angle PCM$ и $\angle PCN$ равны, так как медиана прямоугольного треугольника образует равные углы с катетами относительно биссектрисы угла $C$.
5. По первому признаку равенства треугольников (по двум сторонам и углу между ними) $\triangle CMP = \triangle CNP$. Из равенства треугольников следует равенство их сторон: $PM = PN$. Что и требовалось доказать.

### 📝 Решение пункта (б):
1. Найдем гипотенузу $AB$ по теореме Пифагора:
   $$AB = \sqrt{AC^2 + BC^2} = \sqrt{6^2 + 8^2} = \sqrt{100} = 10$$
2. Найдем радиус вписанной окружности $r$ по формуле для прямоугольного треугольника:
   $$r = \frac{AC + BC - AB}{2} = \frac{6 + 8 - 10}{2} = 2$$
3. Так как $CM = CN = r = 2$, треугольник $MCN$ — прямоугольный ($ \angle C = 90^\circ $) и равнобедренный с катетами, равными 2.
4. Гипотенуза $MN$ этого маленького треугольника находится по теореме Пифагора:
   $$MN = \sqrt{CM^2 + CN^2} = \sqrt{2^2 + 2^2} = \sqrt{8} = 2\sqrt{2}$$

**Ответ: $2\sqrt{2}$**

```python
import matplotlib.pyplot as plt
import numpy as np

def draw_geometry_task(ac, bc):
    # Координаты вершин треугольника ABC (C в начале координат)
    C = np.array([0.0, 0.0])
    A = np.array([0.0, float(ac)])
    B = np.array([float(bc), 0.0])

    # Гипотенуза и радиус вписанной окружности
    ab = np.sqrt(ac**2 + bc**2)
    r = (ac + bc - ab) / 2

    # Центр вписанной окружности O и точки касания M, N
    O = np.array([r, r])
    M = np.array([0.0, r])  # точка касания на AC
    N = np.array([r, 0.0])  # точка касания на BC

    # Середина гипотенузы P
    P = (A + B) / 2

    # Расчет ответа
    mn_length = np.sqrt((M[0] - N[0])**2 + (M[1] - N[1])**2)

    # --- ОТРИСОВКА ---
    plt.figure(figsize=(8, 8))

    # Рисуем треугольник ABC
    plt.plot([C[0], A[0], B[0], C[0]], [C[1], A[1], B[1], C[1]], color='black', lw=2)

    # Рисуем вписанную окружность
    t = np.linspace(0, 2*np.pi, 200)
    plt.plot(O[0] + r*np.cos(t), O[1] + r*np.sin(t), color='blue', linestyle='--', label='Вписанная окружность')

    # Рисуем отрезки PM, PN и MN
    plt.plot([P[0], M[0]], [P[1], M[1]], color='purple', lw=1.5, label='Медианные отрезки PM и PN')
    plt.plot([P[0], N[0]], [P[1], N[1]], color='purple', lw=1.5)
    plt.plot([M[0], N[0]], [M[1], N[1]], color='red', lw=2, label=f'Искомый отрезок MN ({mn_length:.3f})')

    # Отмечаем точки
    points = {'C': C, 'A': A, 'B': B, 'O': O, 'M': M, 'N': N, 'P': P}
    for name, pt in points.items():
        plt.scatter(pt[0], pt[1], color='black', zorder=5)
        plt.text(pt[0] + 0.15, pt[1] + 0.15, name, fontsize=12, fontweight='bold')

    plt.gca().set_aspect('equal')
    plt.grid(True, linestyle=':', alpha=0.5)
    plt.xlim(-1, bc + 1)
    plt.ylim(-1, ac + 1)
    plt.title(f'Точный чертеж к Заданию 16 (Катеты: {ac} и {bc})', fontsize=13)
    plt.legend(loc='upper right')
    plt.show()

    print("--- Проверка геометрического расчета ---")
    print(f"Гипотенуза AB = {ab:.1f}")
    # Выведем ответ в аналитическом виде 2 * sqrt(2) приблизительно
    print(f"Вычисленная длина отрезка MN = {mn_length:.4f}")
    print(f"Теоретический ответ 2*sqrt(2) = {2 * np.sqrt(2):.4f} -> Совпадение!")

# Запуск со значениями из условия задачи (AC=6, BC=8)
draw_geometry_task(ac=6, bc=8)
```

# 🚀 Задание 17. Сложная стереометрия (Геометрия в пространстве)

## 📋 Пошаговая инструкция (Теория для экзамена)

Один из самых мощных алгоритмов для решения сложных задач стереометрии — **координатно-векторный метод**. Он позволяет полностью уйти от сложных геометрических построений и свести задачу к чистой алгебре.

### 🔹 Алгоритм координатного метода в 3D
1.  **Шаг 1: Ввести систему координат.** Поместить начало координат $(0,0,0)$ в одну из вершин фигуры (например, в нижний левый угол куба или призмы). Направить оси $Ox$, $Oy$ и $Oz$ вдоль ребер.
2.  **Шаг 2: Найти координаты ключевых точек.** Выразить координаты нужных вершин через длины ребер фигуры.
3.  **Шаг 3: Использовать векторы для расчетов.**
    *   **Угол между прямыми:** Если прямые заданы векторами $\vec{a}(x_1;y_1;z_1)$ и $\vec{b}(x_2;y_2;z_2)$, то косинус угла между ними находится по формуле:
        $$\cos \alpha = \frac{|\vec{a} \cdot \vec{b}|}{|\vec{a}| \cdot |\vec{b}|} = \frac{|x_1 x_2 + y_1 y_2 + z_1 z_2|}{\sqrt{x_1^2 + y_1^2 + z_1^2} \cdot \sqrt{x_2^2 + y_2^2 + z_2^2}}$$

## 🎯 Разбор типовой задачи КИМ
**Условие:** В правильной треугольной призме $ABCA_1B_1C_1$ все ребра равны $2$. Точка $M$ — середина ребра $AA_1$.
а) Докажите, что прямые $MB$ и $B_1C$ перпендикулярны.
б) Найдите угол между этими прямыми, если ребро основания изменить на другое (в данном случае ребра равны 2, найдем угол для текущей конфигурации, чтобы доказать перпендикулярность, то есть угол $90^\circ$).

### 📝 Решение пункта (а) координатным методом:
1. Поместим начало координат в точку $A(0,0,0)$.
   * Ось $Ox$ направим вдоль ребра $AB$. Тогда точка $B$ имеет координаты $(2, 0, 0)$.
   * Поскольку треугольник $ABC$ правильный со стороной 2, его высота равна $\sqrt{3}$. Направим ось $Oy$ вдоль высоты. Тогда координата вершины $C$ будет $(1, \sqrt{3}, 0)$.
   * Ось $Oz$ направим вертикально вверх вдоль ребра $AA_1$.
2. Запишем координаты интересующих нас точек:
   * $A(0, 0, 0)$, ребро $AA_1 = 2$. Точка $M$ — середина $AA_1 \implies M(0, 0, 1)$.
   * $B(2, 0, 0)$.
   * $B_1$ находится строго над $B \implies B_1(2, 0, 2)$.
   * $C(1, \sqrt{3}, 0)$.
3. Найдем координаты векторов $\vec{MB}$ и $\vec{B_1C}$ (из конца вычитаем начало):
   * $\vec{MB} = (2 - 0; 0 - 0; 0 - 1) = (2; 0; -1)$
   * $\vec{B_1C} = (1 - 2; \sqrt{3} - 0; 0 - 2) = (-1; \sqrt{3}; -2)$
4. Посчитаем скалярное произведение векторов $\vec{MB} \cdot \vec{B_1C}$:
   $$\vec{MB} \cdot \vec{B_1C} = 2 \cdot (-1) + 0 \cdot \sqrt{3} + (-1) \cdot (-2) = -2 + 0 + 2 = 0$$
5. Так как скалярное произведение векторов равно нулю, векторы (и содержащие их прямые) **перпендикулярны**. Угол между ними равен $90^\circ$. Что и требовалось доказать.

## 📚 Где почитать в учебниках

*   **Геометрия 10–11 класс (авторы: Атанасян Л.С. и др.):**
    *   *Глава V:* «Метод координат в пространстве» (§1 — Координаты точки и координаты вектора, §2 — Скалярное произведение векторов). Полная база для координатного метода решения стереометрии.


```python
import matplotlib.pyplot as plt
import numpy as np

def visualize_stereo_task():
    # Координаты вершин треугольной призмы (все ребра равны 2)
    # Нижнее основание
    A = np.array([0, 0, 0])
    B = np.array([2, 0, 0])
    C = np.array([1, np.sqrt(3), 0])

    # Верхнее основание
    A1 = np.array([0, 0, 2])
    B1 = np.array([2, 0, 2])
    C1 = np.array([1, np.sqrt(3), 2])

    # Точка M - середина AA1
    M = np.array([0, 0, 1])

    # Создаем 3D график
    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(111, projection='3d')

    # Отрисовка ребер нижнего и верхнего оснований
    for base in [[A, B, C, A], [A1, B1, C1, A1]]:
        xs = [p[0] for p in base]
        ys = [p[1] for p in base]
        zs = [p[2] for p in base]
        ax.plot(xs, ys, zs, color='gray', linestyle='-', alpha=0.5)

    # Вертикальные ребра
    for p_low, p_high in zip([A, B, C], [A1, B1, C1]):
        ax.plot([p_low[0], p_high[0]], [p_low[1], p_high[1]], [p_low[2], p_high[2]], color='gray', linestyle='-', alpha=0.5)

    # Рисуем скрещивающиеся прямые из условия
    ax.plot([M[0], B[0]], [M[1], B[1]], [M[2], B[2]], color='red', lw=2.5, label='Прямая MB')
    ax.plot([B1[0], C[0]], [B1[1], C[1]], [B1[2], C[2]], color='blue', lw=2.5, label='Прямая B1C')

    # Подписи вершин
    points = {'A': A, 'B': B, 'C': C, 'A1': A1, 'B1': B1, 'C1': C1, 'M': M}
    for name, pt in points.items():
        ax.scatter(pt[0], pt[1], pt[2], color='black', s=40)
        ax.text(pt[0]*1.05, pt[1]*1.05, pt[2], name, fontsize=11)

    # Настройка осей и отображения
    ax.set_xlabel('Ось X')
    ax.set_ylabel('Ось Y')
    ax.set_zlabel('Ось Z')
    ax.set_title('3D-визуализация задачи 17 (Скрещивающиеся прямые)')
    ax.view_init(elev=20, azim=45)
    ax.legend(loc='upper left')

    plt.show()

    # Алгебраический расчет угла через векторы
    vec_MB = B - M
    vec_B1C = C - B1

    scalar_prod = np.dot(vec_MB, vec_B1C)
    norm_MB = np.linalg.norm(vec_MB)
    norm_B1C = np.linalg.norm(vec_B1C)

    cos_alpha = abs(scalar_prod) / (norm_MB * norm_B1C)
    angle_rad = np.arccos(np.clip(cos_alpha, -1.0, 1.0))
    angle_deg = np.degrees(angle_rad)

    print("--- Векторный расчет в пространстве ---")
    print(f"Вектор MB: {vec_MB}")
    print(f"Вектор B1C: {vec_B1C}")
    print(f"Скалярное произведение: {scalar_prod:.2f}")
    print(f"Угол между прямыми: {angle_deg:.1f}°")

visualize_stereo_task()
```

# 🚀 Задание 18. Задачи с параметрами

Это самое высокооцениваемое алгебраическое задание КИМ (4 первичных балла). Оно требует умения исследовать количество и свойства корней уравнений или систем при различных значениях переменной $a$.

## 📋 Пошаговая инструкция (Теория для экзамена)

Разберем один из наиболее наглядных методов — **метод графического анализа на координатной плоскости $(x; a)$**.

### 🔹 В чем суть метода $(x; a)$?
1. Вместо привычной плоскости $(x; y)$, мы строим графики в осях, где горизонтальная ось — это $x$, а вертикальная ось — это параметр $a$.
2. Мы выражаем параметр $a$ через $x$ из исходного уравнения: $a = f(x)$.
3. Строим график функции $a = f(x)$ на плоскости.
4. Ответ на вопрос задачи («при каких $a$ уравнение имеет $k$ корней») находится с помощью мысленного сканирования графика **горизонтальной прямой линией** (линейкой) при разных уровнях высоты $a$. Количество точек пересечения линейки с графиком — это и есть количество корней.

---

## 🎯 Разбор типовой задачи КИМ
**Условие:** Найдите все значения параметра $a$, при каждом из которых уравнение имеет ровно два различных корня:
$$|x^2 - 4x| = a$$

1. **Шаг 1: Проанализировать функции.**
   Левая часть уравнения — это модуль параболы $a = |x^2 - 4x|$.
   Правая часть — это семейство горизонтальных прямых линий на плоскости $(x; a)$.

2. **Шаг 2: Построить график базовой функции.**
   Вспомним, как строится график параболы с модулем:
   * Сначала строим обычную параболу $y = x^2 - 4x$. Ее корни: $x=0$ и $x=4$. Вершина находится в точке $x_0 = -\frac{b}{2a} = 2$, при этом $y_0 = 2^2 - 4\cdot2 = -4$.
   * Знак модуля $| \dots |$ отражает всю нижнюю часть параболы (которая лежит под осью $Ox$) зеркально вверх. То есть вершина параболы перейдет из точки $(2; -4)$ в точку $(2; 4)$.

3. **Шаг 3: Провести сканирование параметрической «линейкой».**
   Начнем двигать горизонтальную прямую линию снизу вверх (от $-\infty$ до $+\infty$) по оси $a$:
   * При $a < 0$: прямая лежит ниже графика. Пересечений нет $\implies$ **0 корней**.
   * При $a = 0$: прямая касается «ножек» графика в точках $x=0$ и $x=4$ $\implies$ **2 корня** (подходит).
   * При $0 < a < 4$: прямая пересекает график в четырех точках $\implies$ **4 корня**.
   * При $a = 4$: прямая касается вершины отраженной параболы и пересекает внешние ветви $\implies$ **3 корня**.
   * При $a > 4$: прямая пересекает только внешние уходящие вверх ветви параболы $\implies$ **2 корня** (подходит).

4. **Шаг 4: Собрать подходящие промежутки.**
   Ровно два корня уравнение имеет в точке $a = 0$, а также на всем интервале от 4 до $+\infty$.

**Ответ: $a = 0 \cup (4; +\infty)$**

```python
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact, FloatSlider

def dynamic_parameter_scanner(a):
    # Задаем сетку для аргумента x
    x = np.linspace(-2, 6, 1000)
    # Исходная функция под модулем: abs(x^2 - 4x)
    y_graph = np.abs(x**2 - 4*x)

    plt.figure(figsize=(10, 6))

    # Рисуем график функции
    plt.plot(x, y_graph, color='#1f77b4', lw=2.5, label='y = |x² - 4x|')

    # Рисуем горизонтальную линию параметра a
    plt.axhline(a, color='red', lw=2, linestyle='--', label=f'Линия параметра a = {a:.2f}')

    # Вычисляем точное количество корней аналитически для вывода на экран
    # Для уравнения |x^2 - 4x| = a
    roots = []
    if a == 0:
        roots = [0, 4]
    elif 0 < a < 4:
        # Два уравнения: x^2 - 4x = a и x^2 - 4x = -a
        roots = [
            2 - np.sqrt(4 + a), 2 + np.sqrt(4 + a),
            2 - np.sqrt(4 - a), 2 + np.sqrt(4 - a)
        ]
    elif a == 4:
        roots = [2, 2 - np.sqrt(8), 2 + np.sqrt(8)]
    elif a > 4:
        roots = [2 - np.sqrt(4 + a), 2 + np.sqrt(4 + a)]

    # Отмечаем точки пересечения (корни) на графике
    for r in roots:
        plt.scatter(r, a, color='red', s=100, zorder=5, edgecolors='black')
        plt.axvline(r, color='gray', linestyle=':', alpha=0.4)

    # Сетка и оформление осей
    plt.axhline(0, color='black', linewidth=1)
    plt.axvline(0, color='black', linewidth=1)
    plt.grid(True, linestyle=':', alpha=0.6)
    plt.xlim(-2, 6)
    plt.ylim(-1, 7)

    plt.title(f'Графический анализ Задания 18. Обнаружено корней: {len(roots)}', fontsize=13, pad=10)
    plt.xlabel('Ось X')
    plt.ylabel('Ось параметров А')
    plt.legend(loc='upper right')
    plt.show()

    # Печатаем текстовый разбор в консоль
    print(f"--- Результат сканирования при a = {a:.2f} ---")
    print(f"Количество найденных пересечений (корней): {len(roots)}")
    if len(roots) > 0:
        formatted_roots = ", ".join([f"{r:.3f}" for r in roots])
        print(f"Приблизительные значения корней x: {formatted_roots}")
    else:
        print("Действительных корней нет.")

# Создаем интерактивный ползунок для параметра 'a'
interact(
    dynamic_parameter_scanner,
    a=FloatSlider(value=0.0, min=-0.5, max=6.0, step=0.1, description='Параметр a:')
);
```

## 🎯 Разбор типовой задачи КИМ (Про среднее арифметическое)
**Условие:** На доске написано 30 натуральных чисел. Известно, что среднее арифметическое всех этих чисел равно 15. Вместо нескольких чисел (возможно, одного) записали их среднее арифметическое, в результате чего количество чисел уменьшилось, а их среднее арифметическое осталось равным 15.
а) Могло ли изначально на доске быть ровно 10 чисел, равных 15?
б) Могло ли после уменьшения на доске остаться ровно 2 числа?
в) Какое наибольшее количество различных чисел могло изначально быть на доске?

### 📝 Решение пункта (а):
Да, могло. Сумма всех 30 чисел равна $30 \cdot 15 = 450$. Если 10 чисел равны 15, их сумма равна 150. На оставшиеся 20 чисел приходится сумма $450 - 150 = 300$. Мы можем взять, например, десять двоек и десять двадцати восьми: $10 \cdot 2 + 10 \cdot 28 = 20 + 280 = 300$. Все условия выполнены, пример существует.

### 📝 Решение пункта (б):
Да, могло. Операция замены группы чисел на их среднее арифметическое не меняет общую сумму всех чисел на доске. Если в конце осталось 2 числа со средним арифметическим 15, их сумма должна быть равна $2 \cdot 15 = 30$. Но изначально сумма 30 натуральных чисел (минимальное натуральное число — это 1) должна быть как минимум $30 \cdot 1 = 30$. Значит, единственный вариант — если изначально на доске были написаны 30 единиц. Но тогда их среднее арифметическое было бы равно 1, а не 15. Противоречие. Ответ: нет, не могло.

### 📝 Решение пункта (в):
1. **Оценка:** Пусть на доске написано 30 различных натуральных чисел. Чтобы их количество различных значений было максимальным, возьмем самые маленькие уникальные натуральные числа: $1, 2, 3, \dots, 30$.
   Посчитаем их сумму по формуле арифметической прогрессии:
   $$S_{min} = \frac{1 + 30}{2} \cdot 30 = 31 \cdot 15 = 465$$
   Но по условию задачи сумма всех чисел жестко зафиксирована и равна $30 \cdot 15 = 450$. Поскольку минимально возможная сумма 30 различных чисел ($465$) превышает реальную сумму ($450$), все 30 чисел различными быть **не могут**.
   Если различных чисел будет 29, то минимальная сумма (если заменить одно число на дубликат) может уменьшиться. Давайте проверим, сколько различных чисел мы можем набрать, чтобы сумма не превысила 450.
2. **Пример:** Попробуем составить набор, где будет как можно больше уникальных чисел. Возьмем первые 28 уникальных чисел от 1 до 28. Их сумма равна $\frac{1+28}{2}\cdot 28 = 406$.
   Осталось распределить $450 - 406 = 44$ единицы на оставшиеся два числа. Мы можем взять числа 15 и 29 (число 15 уже есть в наборе, число 29 — новое уникальное). Получаем 29 уникальных чисел.

**Ответ: а) Да; б) Нет; в) 29**

```python
import itertools

def find_number_combinations():
    print("--- Запуск численного перебора для Задания 19 ---")

    # Решаем пункт (а): ищем набор из 30 чисел со средней суммой 450, где ровно 10 чисел равны 15
    # Зафиксируем 10 пятнадцаток, осталось распределить сумму 300 на 20 натуральных чисел
    # Чтобы не перебирать все 20 чисел (это долго), попробуем составить остаток из двух групп по 10 одинаковых чисел

    found_a = False
    for x in range(1, 15): # первое число меньше 15
        for y in range(16, 31): # второе число больше 15
            if 10 * x + 10 * y == 300:
                example_set = [15]*10 + [x]*10 + [y]*10
                print(f"[Пункт а]: Пример успешно найден!")
                print(f"Набор содержит 10 чисел '15', 10 чисел '{x}' и 10 чисел '{y}'.")
                print(f"Проверка: Всего чисел = {len(example_set)}, Сумма = {sum(example_set)}, Среднее = {sum(example_set)/30}")
                found_a = True
                break
        if found_a: break

    # Решаем пункт (в): Проверим математически максимальное количество уникальных чисел
    # Посчитаем сумму первых N уникальных чисел и найдем грань, когда сумма превысит 450
    print("\n[Пункт в]: Проверка математической оценки...")
    current_sum = 0
    distinct_count = 0
    for i in range(1, 40):
        current_sum += i
        if current_sum <= 450:
            distinct_count = i
        else:
            print(f"Сумма первых {i} уникальных чисел равна {current_sum} (превышает лимит 450).")
            print(f"Следовательно, максимум уникальных чисел без дубликатов не может превышать {distinct_count}.")
            break

find_number_combinations()
```
